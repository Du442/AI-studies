```python
lista3 = [8.5, 6.0, 9.5, 7.0, 5.5, 0.5]

for i in lista3:
    if i >= 9.0:
        print(f'Excelente nota: {i}')
    elif i >= 7.0:
        print(f'Boa nota: {i}')
    elif i >= 5.0:
        print(f'Nota regular: {i}')
    else:
        print(f'Nota insuficiente: {i}')
```

```python
lista4 = [15, 8, 23, 4, 42, 11, 7, 19]
soma = 0
maior = lista4[0]
menor = lista4[0]
lista_impares = list()
lista_pares = list()

for i in lista4:
    # Soma
    soma += i

    # Maior e menor
    if i > maior:
        maior = i
    if i < menor:
        menor = i

    # Numeros pares e impares
    if i % 2 == 0:
        lista_pares.append(i)
    else:
        lista_impares.append(i)

# Media
media = soma / len(lista4)

print(soma)
print(media)
print(maior)
print(menor)
print(lista_pares)
print(lista_impares)
```
