# connecting IA in my notebook

```python
from dotenv import load_dotenv
import os

load_dotenv()

GOOGLE_API_KEY = os.getenv('GEMINI_API_KEY')
```

```python
from google import genai

client = genai.Client(api_key=GOOGLE_API_KEY)
modelo = 'gemini-2.5-flash'
```

```python
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents='Como a IA funciona?'
)
print(response.text)
```

```python
interaction = client.interactions.create(
    model='gemini-2.5-flash',
    input='Como a IA funciona?',
    generation_config={
        'thinking_level': 'low'
    }
)
print(interaction.output_text)
```

```python
chat = client.chats.create(model=modelo)
```

```python
resposta = chat.send_message('O que é a inteligencia artificial?')
```

```python
print(resposta.text)
```

```python
resposta = chat.send_message('Quando foi lançado o chatgpt?')
print(resposta.text)
```

```python
chat.get_history()
```

```python
def exibir_opcoes():
    print('''
Digite a opção escolhida
1 - Mandar pergunta
2 - Visualizar histórico
3 - Sair
''')

while True:
    exibir_opcoes()
    opcao_escolhida = int(input('\nDigite a opção escolhida: '))
    if opcao_escolhida == 1:
        prompt = input('Digite sua mensagem: ')
        resposta = chat.send_message(prompt)
        print(resposta.text)
    elif opcao_escolhida == 2:
        print(chat.get_history())
    elif opcao_escolhida == 3:
        break
    else:
        print('Digite uma opção válida.\n')
```
